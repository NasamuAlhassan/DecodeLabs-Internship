```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
sns.set_palette("Set2")
plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

print("All libraries loaded.")
```

```python
cities = {
    "Accra":    {"lat": 5.5560,  "lon": -0.1969},
    "Kumasi":   {"lat": 6.6885,  "lon": -1.6244},
    "Tamale":   {"lat": 9.4075,  "lon": -0.8533},
    "Takoradi": {"lat": 4.8845,  "lon": -1.7554},
    "Bawku":    {"lat": 11.0601, "lon": -0.2401},
}

START_DATE = "2024-01-01"
END_DATE   = "2024-12-31"

print(f"Cities: {list(cities.keys())}")
print(f"Period: {START_DATE} → {END_DATE}")
```

```python
BASE_URL = "https://archive-api.open-meteo.com/v1/archive"

DAILY_VARS = [
    "temperature_2m_max",
    "temperature_2m_min",
    "temperature_2m_mean",
    "precipitation_sum",
    "windspeed_10m_max",
    "relative_humidity_2m_mean",
]

def fetch_city_weather(city_name, lat, lon):
    params = {
        "latitude":   lat,
        "longitude":  lon,
        "start_date": START_DATE,
        "end_date":   END_DATE,
        "daily":      DAILY_VARS,
        "timezone":   "Africa/Accra",
    }
    response = requests.get(BASE_URL, params=params)
    response.raise_for_status()
    df = pd.DataFrame(response.json()["daily"])
    df["city"] = city_name
    return df

frames = []
for city, coords in cities.items():
    print(f"  Fetching {city}...", end=" ")
    df = fetch_city_weather(city, coords["lat"], coords["lon"])
    frames.append(df)
    print(f" {len(df)} rows")

raw_df = pd.concat(frames, ignore_index=True)
print(f"\n Total records: {len(raw_df)}")
```

```python
print(f"Shape     : {raw_df.shape}")
print(f"Columns   : {list(raw_df.columns)}")
print(f"\nData Types:")
print(raw_df.dtypes)
print(f"\nFirst 5 rows:")
raw_df.head()
```

```python
# Missing values check
print("Missing values:")
print(raw_df.isnull().sum())
```

```python
# Basic stats
print("Basic statistics:")
raw_df.describe().round(2)
```
